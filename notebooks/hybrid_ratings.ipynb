# Hybrid neural book rating prediction

Explicit ratings only. Evaluation reports known-user/item coverage; cold-start pairs are excluded and are not covered by the reported score.

```python
from pathlib import Path
import json
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow import keras
ROOT = Path.cwd() if (Path.cwd() / "data").exists() else Path.cwd().parent
DATA = ROOT / "data"
OUT = ROOT / "artifacts"
OUT.mkdir(exist_ok=True)
keras.utils.set_random_seed(42)
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error
from tensorflow.keras.models import Model

import matplotlib.pyplot as plt
```

```python
ratings = pd.read_csv(DATA / "Ratings.csv", dtype={"ISBN": str})
# Zero denotes implicit feedback; this experiment predicts explicit ratings only.
ratings = ratings.loc[ratings["Book-Rating"].between(1, 10)].dropna(subset=["User-ID", "ISBN"])
ratings = ratings.groupby(["User-ID", "ISBN"], as_index=False)["Book-Rating"].mean()
train_raw, test_raw = train_test_split(ratings, test_size=0.1, random_state=42)
train_raw, val_raw = train_test_split(train_raw, test_size=0.1, random_state=42)
users_new_id = {u: i+1 for i, u in enumerate(sorted(train_raw["User-ID"].unique()))}
books_new_id = {b: i+1 for i, b in enumerate(sorted(train_raw["ISBN"].unique()))}
def encode(frame):
    known = frame["User-ID"].isin(users_new_id) & frame["ISBN"].isin(books_new_id)
    print(f"Known-user/item coverage: {known.mean():.1%} ({known.sum()}/{len(frame)})")
    frame = frame.loc[known]
    if frame.empty:
        raise ValueError("No known user-item pairs in this partition; use a larger dataset.")
    return pd.DataFrame({0: frame["User-ID"].map(users_new_id), 1: frame["ISBN"].map(books_new_id), 2: frame["Book-Rating"]}).reset_index(drop=True)
train, validation, test = map(encode, [train_raw, val_raw, test_raw])
assert set(zip(train_raw["User-ID"], train_raw.ISBN)).isdisjoint(zip(test_raw["User-ID"], test_raw.ISBN))
```

```python
user_input_tensor = keras.layers.Input(shape=[1],name='user-input')
user_embedding_mlp = keras.layers.Embedding(len(users_new_id)+1, 10, 
                    name='user-embedding-mlp')(user_input_tensor)
user_vector_mlp = keras.layers.Flatten(name='user-flatten-mlp')(user_embedding_mlp) 
```

```python
book_input_tensor = keras.layers.Input(shape=[1],name='book-input')
book_embedding_mlp = keras.layers.Embedding(len(books_new_id)+1, 10, 
                    name='book-embedding')(book_input_tensor)
book_vector_mlp = keras.layers.Flatten(name='book-flatten')(book_embedding_mlp) 
```

```python
print(book_vector_mlp.shape)
print(user_vector_mlp.shape)
```

```python
user_embedding_MF = keras.layers.Embedding(len(users_new_id)+1, 10, 
                    name='user-embedding-mf')(user_input_tensor)
user_vector_MF = keras.layers.Flatten(name='user-flatten-mf')(user_embedding_MF) 

book_embedding_MF = keras.layers.Embedding(len(books_new_id)+1, 10, 
                    name='book-embedding-mf')(book_input_tensor)
book_vector_MF = keras.layers.Flatten(name='book-flatten-mf')(book_embedding_MF) 
```

```python
c = keras.layers.Concatenate(axis=1)([book_vector_mlp, user_vector_mlp])

```

```python
concat_dropout = keras.layers.Dropout(0.2)(c)
dense = keras.layers.Dense(64, name='dense', activation='relu')(concat_dropout)
dense2= keras.layers.Dense(32, name= 'dense_2', activation='relu')(dense)
dense3 = keras.layers.Dense(16, name='dense_3', activation='relu')(dense2)
dense3_bn = keras.layers.BatchNormalization(name='batch-norm-3')(dense3)
drop = keras.layers.Dropout(0.2)(dense3_bn)


pred_mlp = keras.layers.Dense(10, name='pred-mlp', activation='relu')(drop)
pred_mf = keras.layers.Dot(axes=1, normalize=True)([book_vector_MF, user_vector_MF])

concatenate_outputs = keras.layers.Concatenate(axis=1)([pred_mf, pred_mlp])

final_output_layer = keras.layers.Dense(1, activation='relu')(concatenate_outputs)

My_model = Model([user_input_tensor, book_input_tensor], final_output_layer)
```

```python
My_model.compile(loss='mean_absolute_error', optimizer='adam')
```

```python
My_model.summary()
```

```python
tf.config.run_functions_eagerly(False)
fitModel = My_model.fit([train[0], train[1]], train[2], epochs=20, validation_data=([validation[0], validation[1]], validation[2]), callbacks=[keras.callbacks.EarlyStopping(patience=3, restore_best_weights=True)])
```

```python
pd.Series(fitModel.history['loss']).plot(logy=True)
plt.xlabel("Epochs")
plt.ylabel("Train Error")
plt.show()
```

```python
predicted = My_model.predict([test[0], test[1]])
results = {"mae": mean_absolute_error(test[2], predicted), "rmse": float(np.sqrt(mean_squared_error(test[2], predicted))), "train_median_baseline_mae": mean_absolute_error(test[2], np.full(len(test), train[2].median()))}
print(results)
(OUT / "hybrid_metrics.json").write_text(json.dumps(results, indent=2))
```

```python
My_model.save(OUT / "hybrid_model.keras")
(OUT / "hybrid_mappings.json").write_text(json.dumps({"users": {str(k): int(v) for k, v in users_new_id.items()}, "books": books_new_id}, indent=2))
```
